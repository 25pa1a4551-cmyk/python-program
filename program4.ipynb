```python
#that count number of vowels(s)
text = input("enter a string")
vowels = "aeiouAEIOU"
count=0
for char in text:
  if char in vowels:
    count+=1
    print("number of vowels: ",count)
```

```python
# code that prints reverse of the number
def reverse_number(n):
  rev=0
while n > 0:
  digit =n%10
  rev=rev*10+digit
  n=n//10
  return rev
  num=int(input("enter a number:"))
  result=reverse_number(num)
  print("reverse of the number:",result)
```
