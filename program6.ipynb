```python
#tuples operattions
t1=(10, 20, 30, 40)
t2=(50, 60, 70, 80)

#concation
print(t1+t2)

#repeataion
print(t1*2)

#length
print(t1)

#slicing
print(t1[1:])

#membership
if 10 in t1:
  print("yes")
else:
  print("no")

#iteration
for i in t1:
  print(i)
```

```python
#builtin functions
t1=(21,3,28,4)
#len()
print(len(t1))
#sum
print(sum(t1))
#min()
print(max(t1))
#max()
print(max(t1))
#sorted()
print(sorted(t1))
```

```python
#methods in tuples
t1=(1, 2, 3, 'g' ,'v')
print("count:",t1.count('g'))
print("index:",t1.index('v'))
```

```python
#zip function
t=(1,2,3)
t1=('a','b','c')
print(tuple(zip(t,t1)))
```
